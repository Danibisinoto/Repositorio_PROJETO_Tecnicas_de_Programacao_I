```python
import pandas as pd
```

```python
#lendo o arquivo CSV
livros = pd.read_csv("Kindle_data-v2.csv")

#qtde de linhas e colunas do DataFrame
livros.shape
```

```python
livros
```

```python
#lendo as 28 primeiras linhas do DataFrame
livros.head(28)
```

```python
#lendo as 28 últimas linhas do DataFrame
livros.tail(28)
```

```python
#informações sobre o DataFrame, tipos de dados,  valores não nulos e memória utilizada
livros.info()

## >>> demonstrou que as colunas autor, vendedor e data da publicação possui valores nulos, 
# concentrando a maior diferença na coluna data da publicação.
```

```python
#qtde de valores nulos por coluna

livros.isnull().sum()

## >>> confirma a análise anterior: qtde de valores nulos em autor, vendedor e data da publicação.
```

```python
#percentual de nulos em comparação com o total de linhas do DataFrame

(livros.isnull().sum()/livros.shape[0])*100

## >> comparando com o DataFrame, o percentual de valores nulos em autor e vendedor são baixos,
# mas em data da publicação tem percentual significativo, porém não impacta diretamente na análise.
```

```python
#dataframe possui campos com link's não necessários para a análise

livros[['imgUrl','productURL']]
```

```python
# deletando as colunas imgUrl e productURL do DataFrame.

livros.drop(columns = ['imgUrl','productURL'], inplace = True)
```

```python
livros.shape
```

```python
#qtde de livros por categoria (28 categorias com mais livros)

livros[['category_name']].value_counts().head(28)
```

```python
#qtde de livros por autor (28 autores com mais livros)

livros[['author']].value_counts().head(28)
```

```python
#qtde de livros que cada vendedor vendeu (28 maiores vendedores)

livros[['soldBy']].value_counts().head(28)
```

```python
# valor arrecadado com a venda dos livros

livros[['price']].sum()
```

```python
# valor total dos livros vendidos por cada vendedor, ordenado do maior para o menor (28 maiores vendedores)

livros.groupby(["soldBy"])["price"].sum().sort_values(ascending=False).head(28)
```

```python
#qtde de livros que são best-sellers e que não são best-sellers

livros['isBestSeller'].value_counts()
```

```python
# valor arrecadado por livros que são best-sellers e que não são best-sellers

livros.groupby(["isBestSeller"])["price"].sum()
```

```python
#qtde de livros que são best-sellers de autor (28 autores com mais livros best-sellers)

bestseller = livros[livros['isBestSeller'] == True]['author'].value_counts().head(28)
bestseller
```

```python
#qtde de livros que não são best-sellers de autor (28 autores com mais livros que não são best-sellers)

naobestseller = livros[~livros['isBestSeller']]['author'].value_counts().head(28)
naobestseller
```

```python
# média e valor total arrecadado por livros que são best-sellers e que não são best-sellers

livros.groupby(["isBestSeller"])["price"].agg(["mean", "sum"]).round(2)
```

```python
# média e valor total arrecadado por livros vendidos por cada vendedor, 
# ordenado pelo valor total (28 maiores vendedores)

livros.groupby(["soldBy"])["price"].agg(["mean", "sum"]).sort_values("sum", ascending=False).round(2).head(28)
```

```python
# média e valor total arrecadado por livros vendidos de cada autor,
# ordenado pelo valor total (28 maiores autores)

livros.groupby(["author"])["price"].agg(["mean", "sum"]).sort_values("sum", ascending=False).round(2).head(28)
```
